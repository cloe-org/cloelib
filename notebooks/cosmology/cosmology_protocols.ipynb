```python
H0 =  67.7
h = H0/100.
As = 2e-9
Omm = 0.3097
Omc = 0.26067408193677466
Omb = 0.04902591806322532
w = -1
wa = 0
zs = np.linspace(0, 3, 100)

background = CAMBBackground(H0=H0, Omc=Omc, Omb=Omb, w=w, wa=wa, Omk = 0.0, ns = 0.96, As = As, gamma_MG = 0.545)
linear_perturbations = CAMBLinearPerturbations(background, zs)
k_values, z_values, pk_values = linear_perturbations.matter_power_spectrum()
```

```python
nonlinear_perturbations = CAMBNonLinearPerturbations(background, zs, nonlinear_model = 'mead')
k_values_nonlinear, z_values_nonlinear, pk_values_nonlinear = nonlinear_perturbations.matter_power_spectrum()
plt.loglog(k_values, pk_values[0, :]);
plt.loglog(k_values_nonlinear, pk_values_nonlinear[0, :]);
plt.xlabel('k');
plt.ylabel('P(k)');
```
