```python
def validate_surface_profiles(profile, file_profile_suffix, title): 

    clusters.profile = profile
    clusters.overdensity = 200

    # Compute Sigma
    file = "SigmaR_"+file_profile_suffix+".txt"
    
    rad_S = np.genfromtxt(os.path.join(truth_dir,file), usecols=(0), dtype=None) * clusters.h # Mpc/h

    Sigma_200c_tr = np.genfromtxt(os.path.join(truth_dir,file), usecols=(1), dtype=None)/clusters.h    # Ms pc^{-2} h
    Sigma_200m_tr = np.genfromtxt(os.path.join(truth_dir,file), usecols=(2), dtype=None)/clusters.h    # Ms pc^{-2} h
    Sigma_vir_tr  = np.genfromtxt(os.path.join(truth_dir,file), usecols=(3), dtype=None)/clusters.h    # Ms pc^{-2} h

    clusters.overdensity_type = 'crit'
    Sigma_200c = clusters.surface_mass_density(rad_S, z, conc, mass)[0]
    clusters.overdensity_type = 'mean'
    Sigma_200m = clusters.surface_mass_density(rad_S, z, conc, mass)[0]
    clusters.overdensity_type = 'vir'
    Sigma_vir = clusters.surface_mass_density(rad_S, z, conc, mass)[0]

    # Compute DeltaSigma
    file = "DeltaSigmaR_"+file_profile_suffix+".txt"
    
    rad_DS = np.genfromtxt(os.path.join(truth_dir,file), usecols=(0), dtype=None) * clusters.h # Mpc/h

    DeltaSigma_200c_tr = np.genfromtxt(os.path.join(truth_dir,file), usecols=(1), dtype=None)/clusters.h    # Ms pc^{-2} h
    DeltaSigma_200m_tr = np.genfromtxt(os.path.join(truth_dir,file), usecols=(2), dtype=None)/clusters.h    # Ms pc^{-2} h
    DeltaSigma_vir_tr  = np.genfromtxt(os.path.join(truth_dir,file), usecols=(3), dtype=None)/clusters.h    # Ms pc^{-2} h

    clusters.overdensity_type = 'crit'
    DeltaSigma_200c = clusters.excess_surface_mass_density(rad_DS, z, conc, mass)[0]
    clusters.overdensity_type = 'mean'
    DeltaSigma_200m = clusters.excess_surface_mass_density(rad_DS, z, conc, mass)[0]
    clusters.overdensity_type = 'vir'
    DeltaSigma_vir = clusters.excess_surface_mass_density(rad_DS, z, conc, mass)[0]

    # Plot
    fig, ax = plt.subplots(1, 2, figsize=(14,6), gridspec_kw={"wspace": 0.3})

    for i in range(len(ax)):
        ax[i].xaxis.set_tick_params(direction='in', which='both', labelsize=14)
        ax[i].yaxis.set_tick_params(direction='in', which='both', labelsize=14)
        ax[i].set_xlabel(r"$R$ [Mpc/h]", fontsize=18)
        ax[i].set_ylabel("(CLOE - COMB-CL) / COMB-CL * 100", fontsize=16)
        ax[i].set_xscale("log")

    ax[0].set_title(r"$\Sigma(R)$ [h M$_\odot$/pc$^2$]", fontsize=15)
    ax[1].set_title(r"$\Delta\Sigma(R)$ [h M$_\odot$/pc$^2$]", fontsize=15)

    ax[0].plot(rad_S, (Sigma_200c-Sigma_200c_tr)/Sigma_200c_tr*100., label=r"$\Delta_{\rm c}=200$")
    ax[0].plot(rad_S, (Sigma_200m-Sigma_200m_tr)/Sigma_200m_tr*100., label=r"$\Delta_{\rm m}=200$")
    ax[0].plot(rad_S, (Sigma_vir-Sigma_vir_tr)/Sigma_vir_tr*100., label=r"$\Delta_{\rm vir}$")

    ax[1].plot(rad_DS, (DeltaSigma_200c-DeltaSigma_200c_tr)/DeltaSigma_200c_tr*100., label=r"$\Delta_{\rm c}=200$")
    ax[1].plot(rad_DS, (DeltaSigma_200m-DeltaSigma_200m_tr)/DeltaSigma_200m_tr*100., label=r"$\Delta_{\rm m}=200$")
    ax[1].plot(rad_DS, (DeltaSigma_vir-DeltaSigma_vir_tr)/DeltaSigma_vir_tr*100., label=r"$\Delta_{\rm vir}$")

    fig.suptitle(title, fontsize=18)
    ax[0].legend(loc="upper right", fontsize=16)
    ax[1].legend(loc="upper right", fontsize=16)
    plt.show()
    
# »»»»»»»»»»»»»»»»»»»»»»»»»»»»»

profile = "NFW"
file_profile_suffix = "NFW"
title = "NFW profile"
validate_surface_profiles(profile, file_profile_suffix, title)

profile = "BMO"
file_profile_suffix = "NFW_trunc"
title = "BMO profile"
validate_surface_profiles(profile, file_profile_suffix, title)
```
