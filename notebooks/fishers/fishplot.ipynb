```python
from chainconsumer import ChainConsumer
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
jcFish = np.load('./jaxcosmoFish.npy')
```

```python
jcFish
```

```python
inv_jcFish = np.linalg.inv(jcFish)
```

```python
import seaborn as sns
snscolors=sns.color_palette("colorblind")
usercolors = [
           '#8338ec',
              '#3a86ff',
           '#fb5607',
           '#ffbe0b']
```

```python
c = ChainConsumer()
names = ['jaxFish_cloelite']
parsnames = [r'$h$', r'$\Omega_c$', r'$\Omega_b$', r'$\sigma_8$', '$w$', r'$w_a$']
print(parsnames)
fiducial=[ 0.677 , 0.26067408,  0.04902592,  0.8277,     -1.  ,0.        ]
flabs = [r'JAX_Fish_cloelite']
for ii, ff in enumerate(names):
    c.add_covariance(fiducial, inv_jcFish,
                     name=flabs[ii], parameters=parsnames)
```

```python
c.configure(plot_hists=True, sigma2d=False, smooth=4, colors=usercolors, linewidths=2.5, 
            legend_kwargs={"loc": (1.5,1), "fontsize": 20}, shade=[False,False, True, False, True], shade_alpha=0.3,
            legend_color_text=True, legend_location=(1, 0), diagonal_tick_labels=True,
            tick_font_size=12, label_font_size=20, max_ticks=2, bar_shade=True)

fig = c.plotter.plot(parameters=parsnames, figsize="page",legend=True)
```
